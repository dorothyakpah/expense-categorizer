```python
def categorize_expense(description):
    description = description.lower().strip()

    if "uber" in description or "bolt" in description:
        return "Transport"

    elif "rice" in description or "bread" in description:
        return "Groceries"

    elif "netflix" in description or "cinema" in description:
        return "Entertainment"

    elif "electricity" in description or "airtime" in description:
        return "Bills"
    else:
        return "Other"
print(categorize_expense("Uber ride"))
print(categorize_expense("Bought rice"))
print(categorize_expense("Netflix subscription"))
print(categorize_expense("Electricity bill"))
print(categorize_expense("Bought a notebook"))
expense= {
    "item": "Bought rice",
    "amount": 3500,
    "category": "Groceries"
}
print(expense["item"])
print(expense["amount"])
print(expense["category"])
expenses=[]
print(expenses)
expense2= {
    "item": "Uber ride",
    "amount": 5000,
    "category": "Transport"
}
print(expenses)
def add_expense(description, amount):
    try:
        amount = float(amount)

        if not description.strip():
            print("Error: Description cannot be empty.")
            return

        if amount <= 0:
            print("Error: Amount must be greater than zero.")
            return

        category = categorize_expense(description)

        expense = {
            "item": description,
            "amount": amount,
            "category": category
        }

        expenses.append(expense)
        print("Expense added successfully!")

    except ValueError:
        print("Error: Please enter a valid number.")
add_expense("Uber ride", 5000)
add_expense("Bought rice", 3500)
add_expense("Electricity bill", 12000)

print(expenses)
def calculate_total():
    amounts = [expense["amount"] for expense in expenses]
    return sum(amounts)
print(calculate_total())
print(expenses)
```

```python
def display_expenses():
    if not expenses:
        print("No expenses recorded yet.")
        return

    print(f"{'ITEM':<25} {'AMOUNT':<15} {'CATEGORY'}")
    print("-" * 55)

    for expense in expenses:
        print(
            f"{expense['item']:<25} "
            f"{expense['amount']:<15.2f} "
            f"{expense['category']}"
        )
display_expenses()
```

```python
def category_summary():
    summary = {}

    for expense in expenses:
        category = expense["category"]
        amount = expense["amount"]

        if category in summary:
            summary[category] += amount
        else:
            summary[category] = amount

    for category, total in summary.items():
        print(f"{category}: ₦{total:,.2f}")
category_summary()
```

```python
import csv

def export_expenses():
    with open("expenses.csv", "w", newline="") as file:
        writer = csv.DictWriter(
            file,
            fieldnames=["item", "amount", "category"]
        )

        writer.writeheader()
        writer.writerows(expenses)

    print("Expenses exported successfully!")
export_expenses()
```

```python
def run_expense_tracker():
    print("PERSONAL EXPENSE TRACKER")
    print("Enter 'done' when you want to finish.")

    while True:
        description = input("\nEnter expense description: ")

        if description.strip().lower() == "done":
            break

        amount = input("Enter expense amount: ")

        add_expense(description, amount)

    print("\nYour expenses have been recorded!")
    display_expenses()
    category_summary()
    print(f"\nTotal spending: ₦{calculate_total():,.2f}")
run_expense_tracker()
```
