```python
def categorize_expense(description):
    description = description.lower().strip()

    if "uber" in description or "bolt" in description:
        return "Transport"

    elif "rice" in description or "bread" in description:
        return "Groceries"

    elif "netflix" in description or "cinema" in description:
        return "Entertainment"

    elif "electricity" in description or "airtime" in description:
        return "Bills"
    else:
        return "Other"
print(categorize_expense("Uber ride"))
print(categorize_expense("Bought rice"))
print(categorize_expense("Netflix subscription"))
print(categorize_expense("Electricity bill"))
print(categorize_expense("Bought a notebook"))
expense= {
    "item": "Bought rice",
    "amount": 3500,
    "category": "Groceries"
}
print(expense["item"])
print(expense["amount"])
print(expense["category"])
expenses=[]
expenses.append(expense)
print(expenses)
expense2= {
    "item": "Uber ride",
    "amount": 5000,
    "category": "Transport"
}
expenses.append(expense2)
print(expenses)
def add_expense(description, amount):
    category = categorize_expense(description)

    expense = {
        "item": description,
        "amount": amount,
        "category": category
    }

    expenses.append(expense)
add_expense("Uber ride", 5000)
add_expense("Bought rice", 3500)
add_expense("Electricity bill", 12000)

print(expenses)
def calculate_total():
    amounts = [expense["amount"] for expense in expenses]
    return sum(amounts)
print(calculate_total())
print(expenses)
```
