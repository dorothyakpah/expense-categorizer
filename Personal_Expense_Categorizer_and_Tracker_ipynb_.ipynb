# Personal Expense Categorizer and Tracker

A Python application for recording, categorising, analysing, and managing personal expenses.

## Project Overview

This project demonstrates how Python can be used to build a practical expense-management application with automated categorisation, input validation, spending summaries, monthly budget tracking, and data storage using Google Drive.

1. Expense Setup and Categorisation

This section initialises expense storage and defines the categories and automatic categorisation rules used throughout the application.

```python

expenses = []

categories = {
    "Groceries": [
        "rice", "bread", "sugar", "beans", "tomatoes",
        "milk", "eggs", "garri", "yam", "vegetables"
    ],
    "Transport": [
        "uber", "bolt", "petrol", "fuel", "bus", "taxi",
        "train", "flight"
    ],
    "Entertainment": [
        "netflix", "cinema", "movie", "concert", "games"
    ],
    "Bills & Utilities": [
        "electricity", "airtime", "water bill", "internet",
        "gas bill", "waste disposal"
    ],
    "Shopping": [
        "clothes", "shoes", "bag", "watch", "accessories"
    ],
    "Healthcare": [
        "medicine", "medication", "hospital", "doctor",
        "pharmacy", "consultation"
    ],
    "Education": [
        "books", "course", "tuition", "school supplies",
        "textbook"
    ],
    "Personal Care": [
        "skincare", "hair cream", "shampoo", "conditioner",
        "soap", "toothpaste", "perfume"
    ],
    "Appliances & Electronics": [
        "rice cooker", "washing machine", "refrigerator",
        "laptop", "phone", "television", "charger"
    ]
}


def categorize_expense(description):
    description = description.lower().strip()

    matched_keywords = []

    for category, keywords in categories.items():
        for keyword in keywords:
            if keyword in description:
                matched_keywords.append((keyword, category))

    if matched_keywords:
        matched_keywords.sort(
            key=lambda match: len(match[0]),
            reverse=True
        )
        return matched_keywords[0][1]

    return "Other"
```

2. Expense Recording and Validation

This section handles expense entries, validates user input, automatically assigns categories, and stores valid expenses.

```python

import math
from datetime import date, datetime

def add_expense(description, amount, expense_date=None):
    # Validate description
    if not isinstance(description, str):
        print("Error: Description must be text.")
        return

    description = description.strip()

    if not description:
        print("Error: Description cannot be empty.")
        return

    # Validate amount
    if isinstance(amount, bool):
        print("Error: Please enter a valid number.")
        return

    try:
        amount = float(amount)

        if not math.isfinite(amount):
            print("Error: Amount must be a finite number.")
            return

        if amount <= 0:
            print("Error: Amount must be greater than zero.")
            return

    except (ValueError, TypeError):
        print("Error: Please enter a valid number.")
        return

    # Validate expense date
    if expense_date is None:
        expense_date = date.today()
    else:
        try:
            expense_date = datetime.strptime(
                expense_date, "%Y-%m-%d"
            ).date()

        except (ValueError, TypeError):
            print("Error: Please enter a valid date (YYYY-MM-DD).")
            return

    # Categorise and save expense
    category = categorize_expense(description)

    expense = {
        "item": description,
        "amount": amount,
        "category": category,
        "date": expense_date.isoformat()
    }

    expenses.append(expense)

    print("Expense added successfully!")

    # Automatically save data
    save_data()
```

```python
expenses.clear()

# Valid entries
add_expense("Bought rice", 3500)
add_expense("Bought bread", 1500, "2026-11-10")
add_expense("Uber ride", "5000")

# Invalid descriptions
add_expense("", 2000)
add_expense(3500, 2000)

# Invalid amounts
add_expense("Bought shoes", 0)
add_expense("Bought milk", -500)
add_expense("Bought sugar", "abc")
add_expense("Bought eggs", float("nan"))
add_expense("Bought yam", float("inf"))
add_expense("Bought beans", True)

# Invalid dates
add_expense("Bought eggs", 1200, "2026-02-30")
add_expense("Bought milk", 800, "wrong-date")

print("\nFinal expense records:")
print(expenses)
view_expenses()
```

## 3. Viewing Recorded Expenses
This section displays recorded expenses in a clear, organised format, including their descriptions, amounts, and categories.

```python
def view_expenses():
    if not expenses:
        print("No expenses recorded yet.")
        return

    print("\nYOUR EXPENSES")
    print("-" * 95)
    print(
        f"{'No.':<5}"
        f"{'Date':<15}"
        f"{'Description':<25}"
        f"{'Amount (NGN)':<20}"
        f"{'Category'}"
    )
    print("-" * 95)

    for number, expense in enumerate(expenses, start=1):
        amount = f"₦{expense['amount']:,.2f}"

        print(
            f"{number:<5}"
            f"{expense['date']:<15}"
            f"{expense['item']:<25}"
            f"{amount:<20}"
            f"{expense['category']}"
        )

    print("-" * 95)
```

## 4. Spending Calculations
This section calculates total spending and provides useful numerical summaries from recorded expenses.

```python

import math

# Calculate total spending
def calculate_total_expenses():
    return sum(expense["amount"] for expense in expenses)


# Calculate spending by category
def calculate_category_totals():
    category_totals = {}

    for expense in expenses:
        category = expense["category"]
        amount = expense["amount"]

        if category not in category_totals:
            category_totals[category] = 0

        category_totals[category] += amount

    return category_totals


# Display spending summary
def display_spending_summary():
    if not expenses:
        print("No expenses recorded yet.")
        return

    total_count = len(expenses)
    total_amount = calculate_total_expenses()
    category_totals = calculate_category_totals()

    print("\n========== SPENDING SUMMARY ==========")
    print(f"Total expenses recorded: {total_count}")
    print(f"Total amount spent: ₦{total_amount:,.2f}")

    print("\n---------- CATEGORY BREAKDOWN ----------")

    for category, amount in category_totals.items():
        print(f"{category}: ₦{amount:,.2f}")

    print("=======================================")


# Calculate spending for a particular month
def calculate_monthly_expenses(year, month):
    if isinstance(year, bool) or not isinstance(year, int) or year < 1:
        print("Error: Please enter a valid year.")
        return

    if isinstance(month, bool) or not isinstance(month, int) or not 1 <= month <= 12:
        print("Error: Month must be between 1 and 12.")
        return

    monthly_total = 0

    for expense in expenses:
        expense_date = datetime.strptime(
            expense["date"], "%Y-%m-%d"
        ).date()

        if expense_date.year == year and expense_date.month == month:
            monthly_total += expense["amount"]

    return monthly_total


# Set and save a monthly budget
def set_monthly_budget(year, month, amount):
    if isinstance(year, bool) or not isinstance(year, int) or year < 1:
        print("Error: Please enter a valid year.")
        return

    if isinstance(month, bool) or not isinstance(month, int) or not 1 <= month <= 12:
        print("Error: Month must be between 1 and 12.")
        return

    if isinstance(amount, bool):
        print("Error: Please enter a valid budget amount.")
        return

    try:
        amount = float(amount)

        if not math.isfinite(amount):
            print("Error: Budget must be a finite number.")
            return

        if amount <= 0:
            print("Error: Budget must be greater than zero.")
            return

    except (ValueError, TypeError):
        print("Error: Please enter a valid budget amount.")
        return

    budget_key = f"{year}-{month:02d}"
    monthly_budgets[budget_key] = amount

    print(f"Budget set successfully for {budget_key}!")
    save_data()


# Calculate the remaining monthly budget
def calculate_remaining_budget(year, month):
    if isinstance(year, bool) or not isinstance(year, int) or year < 1:
        print("Error: Please enter a valid year.")
        return

    if isinstance(month, bool) or not isinstance(month, int) or not 1 <= month <= 12:
        print("Error: Month must be between 1 and 12.")
        return

    budget_key = f"{year}-{month:02d}"

    if budget_key not in monthly_budgets:
        print("No budget has been set for this month.")
        return

    budget = monthly_budgets[budget_key]
    spent = calculate_monthly_expenses(year, month)

    return budget - spent


# Display monthly budget status
def check_monthly_budget(year, month):
    if isinstance(year, bool) or not isinstance(year, int) or year < 1:
        print("Error: Please enter a valid year.")
        return

    if isinstance(month, bool) or not isinstance(month, int) or not 1 <= month <= 12:
        print("Error: Month must be between 1 and 12.")
        return

    budget_key = f"{year}-{month:02d}"

    if budget_key not in monthly_budgets:
        print("No budget has been set for this month.")
        return

    budget = monthly_budgets[budget_key]
    spent = calculate_monthly_expenses(year, month)
    percentage_used = (spent / budget) * 100

    print(f"\nBUDGET STATUS: {budget_key}")
    print(f"Budget: ₦{budget:,.2f}")
    print(f"Spent: ₦{spent:,.2f}")
    print(f"Budget used: {percentage_used:.1f}%")

    if percentage_used >= 100:
        print("WARNING: You have exceeded your monthly budget!")
    elif percentage_used >= 80:
        print("Warning: You are approaching your monthly budget limit.")
    else:
        print("Your spending is within budget.")
```

5. Editing and Deleting Expenses

```python

def edit_expense(expense_number, new_description=None,
                 new_amount=None, new_date=None):

    # Validate expense number
    if isinstance(expense_number, bool) or not isinstance(expense_number, int):
        print("Error: Please enter a valid expense number.")
        return

    if expense_number < 1 or expense_number > len(expenses):
        print("Error: Expense number does not exist.")
        return

    expense = expenses[expense_number - 1]

    # Validate new description
    if new_description is not None:
        if not isinstance(new_description, str):
            print("Error: Description must be text.")
            return

        new_description = new_description.strip()

        if not new_description:
            print("Error: Description cannot be empty.")
            return

    # Validate new amount
    if new_amount is not None:
        if isinstance(new_amount, bool):
            print("Error: Please enter a valid amount.")
            return

        try:
            new_amount = float(new_amount)

            if not math.isfinite(new_amount):
                print("Error: Amount must be a finite number.")
                return

            if new_amount <= 0:
                print("Error: Amount must be greater than zero.")
                return

        except (ValueError, TypeError):
            print("Error: Please enter a valid amount.")
            return

    # Validate new date
    if new_date is not None:
        if not isinstance(new_date, str):
            print("Error: Date must be text in YYYY-MM-DD format.")
            return

        try:
            parsed_date = datetime.strptime(
                new_date, "%Y-%m-%d"
            ).date()

            if parsed_date.isoformat() != new_date:
                print("Error: Date must use YYYY-MM-DD format.")
                return

        except ValueError:
            print("Error: Please enter a valid date (YYYY-MM-DD).")
            return

    # Check whether changes were requested
    if (new_description is None and
        new_amount is None and
        new_date is None):
        print("No changes were requested.")
        return

    # Apply changes
    if new_description is not None:
        expense["item"] = new_description
        expense["category"] = categorize_expense(new_description)

    if new_amount is not None:
        expense["amount"] = new_amount

    if new_date is not None:
        expense["date"] = new_date

    print("Expense updated successfully!")
    save_data()


def delete_expense(expense_number):

    # Validate expense number
    if isinstance(expense_number, bool) or not isinstance(expense_number, int):
        print("Error: Please enter a valid expense number.")
        return

    if expense_number < 1 or expense_number > len(expenses):
        print("Error: Expense number does not exist.")
        return

    # Delete selected expense
    deleted_expense = expenses.pop(expense_number - 1)

    print(
        f"Expense '{deleted_expense['item']}' "
        "deleted successfully!"
    )

    save_data()
```

 6. Searching and Filtering Expenses

```python
def search_expenses(search_term):
    if not isinstance(search_term, str):
        print("Error: Search term must be text.")
        return

    search_term = search_term.strip().lower()

    if not search_term:
        print("Error: Search term cannot be empty.")
        return

    matching_expenses = []

    for expense in expenses:
        if search_term in expense["item"].lower():
            matching_expenses.append(expense)

    if not matching_expenses:
        print("No matching expenses found.")
        return

    print("\nSEARCH RESULTS")
    print("-" * 75)
    print(
        f"{'Date':<15}"
        f"{'Description':<25}"
        f"{'Amount (NGN)':<20}"
        f"{'Category'}"
    )
    print("-" * 75)

    for expense in matching_expenses:
        amount = f"₦{expense['amount']:,.2f}"

        print(
            f"{expense['date']:<15}"
            f"{expense['item']:<25}"
            f"{amount:<20}"
            f"{expense['category']}"
        )

    print("-" * 75)
```

```python
def filter_expenses_by_category(category):
    if not isinstance(category, str):
        print("Error: Category must be text.")
        return

    category = category.strip().lower()

    if not category:
        print("Error: Category cannot be empty.")
        return

    matching_expenses = [
        expense for expense in expenses
        if expense["category"].lower() == category
    ]

    if not matching_expenses:
        print("No expenses found in this category.")
        return

    print(f"\nEXPENSES IN {category.upper()}")
    print("-" * 75)
    print(
        f"{'Date':<15}"
        f"{'Description':<25}"
        f"{'Amount (NGN)':<20}"
        f"{'Category'}"
    )
    print("-" * 75)

    for expense in matching_expenses:
        amount = f"₦{expense['amount']:,.2f}"

        print(
            f"{expense['date']:<15}"
            f"{expense['item']:<25}"
            f"{amount:<20}"
            f"{expense['category']}"
        )

    print("-" * 75)
```

7. Exporting Expense Records

```python

import csv
import os

def export_expenses(filename=None):

    # Save the CSV file in the project folder by default
    if filename is None:
        filename = os.path.join(DATA_DIR, "expenses.csv")

    fieldnames = ["date", "item", "amount", "category"]

    try:
        with open(filename, mode="w", newline="", encoding="utf-8-sig") as file:
            writer = csv.DictWriter(file, fieldnames=fieldnames)

            writer.writeheader()
            writer.writerows(expenses)

        print(f"Expenses exported successfully to {filename}!")

    except OSError:
        print("Error: Unable to create the export file.")
```

 8. Data Persistence and Storage

```python

from google.colab import drive
import os
import json

# Connect Google Drive
drive.mount('/content/drive')

# Create the project storage folder
DATA_DIR = "/content/drive/MyDrive/Personal_Expense_Tracker"
os.makedirs(DATA_DIR, exist_ok=True)

# Define storage file paths
EXPENSES_FILE = os.path.join(DATA_DIR, "expenses.json")
BUDGETS_FILE = os.path.join(DATA_DIR, "budgets.json")


# Save application data
def save_data():
    try:
        with open(EXPENSES_FILE, "w", encoding="utf-8") as file:
            json.dump(expenses, file, indent=4)

        with open(BUDGETS_FILE, "w", encoding="utf-8") as file:
            json.dump(monthly_budgets, file, indent=4)

        print("Data saved successfully!")

    except OSError:
        print("Error: Unable to save data.")


# Load previously saved data
def load_data():
    try:
        with open(EXPENSES_FILE, "r", encoding="utf-8") as file:
            saved_expenses = json.load(file)

        with open(BUDGETS_FILE, "r", encoding="utf-8") as file:
            saved_budgets = json.load(file)

        if not isinstance(saved_expenses, list):
            print("Error: Invalid expense data.")
            return

        if not isinstance(saved_budgets, dict):
            print("Error: Invalid budget data.")
            return

        expenses.clear()
        expenses.extend(saved_expenses)

        monthly_budgets.clear()
        monthly_budgets.update(saved_budgets)

        print("Data loaded successfully!")

    except FileNotFoundError:
        print("No saved data files found.")

    except json.JSONDecodeError:
        print("Error: Saved data contains invalid JSON.")

    except OSError:
        print("Error: Unable to load saved data.")
```

9. Application Startup and Data Initialisation

```python

# Initialise application data
expenses = []
monthly_budgets = {}

print("========== PERSONAL EXPENSE TRACKER ==========")
print("Initialising application...")

# Load previously saved data
load_data()

# Display application summary
print("\n========== APPLICATION SUMMARY ==========")

print("Total recorded expenses:", len(expenses))

print(
    f"Total spending: ₦{calculate_total_expenses():,.2f}"
)

print("\nCategory totals:")
print(calculate_category_totals())

print("\nMonthly budgets:")
print(monthly_budgets)

print("\nApplication ready!")
```
